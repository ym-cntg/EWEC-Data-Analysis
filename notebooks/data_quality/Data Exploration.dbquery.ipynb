```sql
SELECT date_trunc('MONTH', DateTime) AS month,
       ROUND(corr(Pv2_AVL, CASE WHEN DateTime < '2026-03-17' THEN PV1_MW ELSE PV2_MW END), 3) AS corr_avl_vs_large_plant,
       ROUND(AVG(Pv2_AVL - CASE WHEN DateTime < '2026-03-17' THEN PV1_MW ELSE PV2_MW END), 1) AS avg_avl_minus_actual,
       ROUND(100 * AVG(CASE WHEN Pv2_AVL IN (1500, 1640) THEN 1 ELSE 0 END), 1) AS pct_at_cap,
       ROUND(100 * AVG(CASE WHEN Pv2_AVL IS NULL THEN 1 ELSE 0 END), 1) AS pct_missing
FROM ewec_dev_powerops.bronze.power_contango_minute_pv
WHERE hour(DateTime) BETWEEN 9 AND 15
  AND abs(PV1_MW) < 5000 AND abs(PV2_MW) < 5000
GROUP BY 1 ORDER BY 1;
```
